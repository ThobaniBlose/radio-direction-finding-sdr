```julia
using FFTW
using LinearAlgebra
using Plots
plotly() # plotyly backend for Plots
```

```julia
j = im
c = 3e8

fs = 20e6
dt = 1/fs

Nsamples = 2^15

t = (0:Nsamples - 1) * dt

t_max = Nsamples * dt

# Simulate a real 1MHz sinusoid and do an FFT and plot it
x = cos.(2 * pi * 1e6 * t)
```

```julia
plot(t,x)
```

```julia
X = fft(x)
plot(abs.(X), title = "FFT of x")
```
